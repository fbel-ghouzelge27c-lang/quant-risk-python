 # Projet 01 – Perte attendue d'un portefeuille de crédit ,Calcul de l'EL (PD × LGD × EAD) sur les 5 segments du stress test de Banque Atlas, analyse de concentration et contrôle des données.

portfolio = [{"Segment":"Large Corporates","EAD":42,"PD":0.012,"LGD":0.4},
             {"Segment":"SMEs","EAD":28,"PD":0.035,"LGD":0.45},
             {"Segment":"Real estate","EAD":9,"PD":0.04,"LGD":0.4},
             {"Segment":"Mortgages","EAD":31,"PD":0.015,"LGD":0.2},
         {"Segment":"Consumer loans","EAD":14,"PD":0.045,"LGD":0.65}]
print(portfolio[0])

```python
def expected_loss(PD,EAD,LGD):
    EL=PD*EAD*LGD
    return EL

for seg in portfolio :
   if seg["Segment"]=="Consumer loans":
      print (seg)
print(expected_loss(seg["PD"],seg["EAD"],seg["LGD"]))  
```

```python
seg=0
for seg in portfolio :
    print(expected_loss(seg["PD"],seg["EAD"],seg["LGD"])) 
    
```

```python
seg=0
for seg in portfolio :
    seg["EL"]=expected_loss(seg["PD"],seg["EAD"],seg["LGD"])
print(portfolio)
    
```

```python
def total_EL(portfolio):
    total = 0
    i = 0
    while i < len(portfolio):
        total += portfolio[i]["EL"]
        i += 1
    return total

resultat = total_EL(portfolio)
print(resultat)
```

```python


for seg in portfolio:
        if seg["EL"]>0.3:
            seg["inter"]="High"
        else:
            seg["inter"]=" "
for seg in portfolio :
    seg["part"]=seg["EL"]/resultat
for seg in portfolio:
   print (f"{seg["Segment"]:>18},{seg["PD"]:>8},{seg["EAD"]:>6},{seg["LGD"]:>8},{seg["EL"]:>8.4f},{seg["inter"]:>8},{seg["part"]:>8.4f}")
```
